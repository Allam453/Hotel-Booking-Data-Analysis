```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('/content/hotel_booking.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.drop(columns=[
    'name',
    'email',
    'phone-number',
    'credit_card'
], inplace=True)
```

```python
clean_df=df.copy()
```

```python
clean_df.drop(columns=[
    'name',
    'email',
    'phone-number',
    'credit_card'
], inplace=True)
```

```python
clean_df.isnull().sum()
```

```python
clean_df.columns=clean_df.columns.str.strip()
clean_df.columns=clean_df.columns.str.lower()
clean_df.columns=clean_df.columns.str.replace(" ", "_")
```

```python
clean_df.duplicated().sum()
```

```python
clean_df.drop_duplicates(inplace=True)
```

```python
clean_df["children"]=clean_df["children"].astype(int)
```

```python
clean_df["agent"]=clean_df["agent"].fillna(0)
```

```python
clean_df["agent"]=clean_df["agent"].astype(int)
```

```python
clean_df["country"]=clean_df["country"].fillna("UnKnown")
```

```python
clean_df["meal"]=clean_df["meal"].replace("Undefined", "UnKnown")
```

```python
clean_df.drop(columns='company', inplace=True)
```

```python
clean_df['reservation_status_date']=pd.to_datetime(clean_df['reservation_status_date'])
```

```python
clean_df=clean_df[clean_df['children']<=6]
```

```python
clean_df=clean_df[clean_df['babies']<=6]
```

```python
clean_df=clean_df[clean_df['adr']>=0]
```

```python
clean_df.isnull().sum()
```

```python
clean_df['invalid_guest_count'] = ((clean_df['adults']==0) &
                             (clean_df['children']==0) &
                             (clean_df['babies']==0))
```

```python
clean_df.to_csv('clean_hotel_booking.csv', index=False)
```

```python
clean_df['total_guests'] = clean_df['adults'] + clean_df['children'] + clean_df['babies']
```

```python
clean_df['total_stay_nights'] = clean_df['stays_in_weekend_nights'] + clean_df['stays_in_week_nights']
```

```python
clean_df['booking_status'] = clean_df['is_canceled'].map({
    0:'Not Canceled',
    1:'Canceled'
})
```

```python
clean_df['revenue'] = clean_df['adr'] * clean_df['total_stay_nights']
```

```python
clean_df['year'] = clean_df['reservation_status_date'].dt.year
clean_df['month'] = clean_df['reservation_status_date'].dt.month_name()
```

```python
clean_df['agent'] = clean_df['agent'].astype(str)
```

```python
clean_df['lead_time_category'] = pd.cut(
    clean_df['lead_time'],
    bins=[-1,30,90,180,1000],
    labels=['Short','Medium','Long','Very Long']
)
```

```python
clean_df['stay_category'] = pd.cut(
    clean_df['total_stay_nights'],
    bins=[0,3,7,100],
    labels=['Short Stay','Medium Stay','Long Stay']
)
```

```python
clean_df['family_booking'] = np.where(
    (clean_df['children'] + clean_df['babies']) > 0,
    'Family',
    'No Family'
)
```

```python
clean_df.to_csv('clean_hotel_booking.csv', index=False)
```
